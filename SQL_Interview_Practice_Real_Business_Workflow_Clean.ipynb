# Step 12 — Aggregating Customers by City Using GROUP BY

```python
query = """
SELECT city,count(customer_id) as cnt
FROM customers
group by city
order by cnt desc

"""

pd.read_sql(query,conn)
```

```python
query = """
SELECT *
FROM orders
LEFT JOIN customers
on customers.customer_id = orders.customer_id 

"""

pd.read_sql(query, conn)
```

# Step 13 — Joining Customers and Orders to Analyse Loyalty Members

```python
query = """
SELECT loyalty_member,
    count(order_id)
FROM orders
LEFT JOIN customers
ON orders.customer_id = customers.customer_id
group by loyalty_member 
order by loyalty_member desc
"""

pd.read_sql(query, conn)
```

# Step 14 — Revenue Analysis Using SUM and AVG

```python
query = """
SELECT category,
    AVG(price) AS AvgPrice ,
    SUM(quantity) AS TotalNumber, 
    SUM(price*quantity) AS TotalPay
FROM orders
LEFT JOIN products
ON orders.product_id = products.product_id
GROUP BY category
ORDER BY TotalPay desc
"""

pd.read_sql(query,conn)

```

```python
query = """
SELECT 
    category AS Category,
    SUM((price*quantity)) AS TotalPay,
    AVG(price) AS AvgPrice,
    SUM(quantity) AS TotalNum
FROM orders
LEFT JOIN products
ON products.product_id = orders.product_id
GROUP BY category
HAVING Totalpay >= 300000
ORDER BY TotalPay desc
"""

pd.read_sql(query, conn)
```

# Step 16 — Customer Segmentation Using CASE WHEN

```python
query = """
SELECT 
    *,
    CASE   
    WHEN age<30 THEN "Young"
    WHEN age between 30 and 50 THEN "Adult"
    WHEN age> 50 THEN "Senior"
    END AS Segmentation
FROM customers

"""

pd.read_sql(query, conn)
```
